## Diseño de PCB

El diseño de la placa se llevó a cabo en KiCad tomando como base el esquemático corregido. Para cada componente se seleccionaron los footprints correspondientes a partir de las bibliotecas estándar del software.

### Criterio de diseño adoptados:
- Montaje del fusible (F1): Se le asignó una tira de pines hembra de 1x02 para permitir su reemplazo rápido sin necesidad de desoldar componentes ante una eventual falla.
- Ubicación de periféricos: Los servomotores, joysticks y pulsadores se montarán de forma externa a la placa de expansión (poncho), vinculándose mediante arneses de cables. Esta disposición optimiza el área útil de la PCB y mejora la ergonomía durante la operación de los controles.
