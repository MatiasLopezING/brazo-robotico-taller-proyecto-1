## Modificaciones de conexiones en el diseño esquemático

Previo al diseño de la PCB, se realizaron ajustes en el ruteo y asignación de pines en las terminales de conexión, tal como se ilustra a continuación:

![Circuito esquemático](esquematico_corregido.jpg)
### Figura 1. Circuito esquematico modificado.

### Cambios realizados en el conector P1:

* **Reasignación de la 4.ª consigna analógica (`JOY_B_Y`):** En la versión preliminar, la señal analógica correspondiente al eje Y del Joystick B se encontraba asignada al pin 15. En la versión definitiva se trasladó al **Pin 35 de P1**.
* **Liberación del Pin 15:** El pin 15 (asociado a la salida DAC/CH0) se desconectó de la línea de mando y quedó marcado formalmente sin conexión (*No Connect*).
* **Reubicación de la salida PWM de la Muñeca:** El **Pin 35**, que originalmente portaba la modulación `PWM_MUÑECA`, cedió su posición al ingreso de `JOY_B_Y` al migrar dicha consigna PWM al conector P2.
* **Reubicación de la salida PWM de la Base:** El **Pin 36**, que comandaba la señal `PWM_BASE`, se desvinculó de P1 y se marcó sin conexión tras trasladar el control de la base a P2.


### Cambios realizados en el conector P2:

* **Corrección del pulsador de la pinza (`PINZA_BTN`):** En el diseño original, la entrada digital del botón se conectaba al pin 31 en proximidad a la línea de tierra. En la versión corregida, la señal `PINZA_BTN` se trasladó de forma independiente al **Pin 29 de P2** (`GPIO0`), evitando interferencias con el nodo de masa.
* **Incorporación de `PWM_MUÑECA`:** La salida de control para el servomotor de la muñeca se reasignó al **Pin 31 de P2** (`SERVO8` / `PWM10`), ocupando la posición liberada por el pulsador.
* **Incorporación de `PWM_BASE`:** La modulación para el servomotor de la base se trasladó desde el pin 36 de P1 hacia el **Pin 35 de P2** (`GPIO6`).

## Diseño de la PCB

El diseño de la placa se llevó a cabo en KiCad tomando como base el esquemático corregido. Para cada componente se seleccionaron los footprints correspondientes a partir de las bibliotecas estándar del software.

### Criterio de diseño adoptados:
- **Montaje del fusible (F1):** Se le asignó una tira de pines hembra de 1x02 para permitir su reemplazo rápido sin necesidad de desoldar componentes ante una eventual falla.
- **Ubicación de periféricos:** Los servomotores, joysticks y pulsadores se montarán de forma externa a la placa de expansión (poncho), vinculándose mediante arneses de cables. Esta disposición optimiza el área útil de la PCB y mejora la ergonomía durante la operación de los controles.
